```python
import hashlib
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
np.set_printoptions(suppress=True)

def check(label, answer, key, d=2):
    """Compare your answer with the hidden key (rounded to d decimals)."""
    if answer is None:
        print(f"⏳ {label}: not answered yet · មិនទាន់បានឆ្លើយ")
        return
    try:
        arr = np.atleast_1d(np.asarray(answer, dtype=float)).ravel()
        s = "|".join(f"{(0.0 if abs(v) < 0.5 * 10**-d else v):.{d}f}" for v in arr)
        ok = hashlib.sha256(s.encode()).hexdigest()[:16] == key
    except Exception as e:
        print(f"❌ {label}: could not read your answer ({e})")
        return
    if ok:
        print(f"✅ {label}: correct · ត្រឹមត្រូវ")
    else:
        print(f"❌ {label}: not yet, try again · មិនទាន់ត្រូវ សាកល្បងម្តងទៀត")

def check_between(label, answer, low, high):
    """For simulations: the answer only needs to fall in a range."""
    if answer is None:
        print(f"⏳ {label}: not answered yet · មិនទាន់បានឆ្លើយ")
    elif low <= float(answer) <= high:
        print(f"✅ {label}: {float(answer):.4f} is in [{low}, {high}] · ត្រឹមត្រូវ")
    else:
        print(f"❌ {label}: {float(answer):.4f} is outside [{low}, {high}] · សាកល្បងម្តងទៀត")

# The student dataset used in Chapters 5-7 of the book (same random seed)
rng = np.random.default_rng(7)
hours = rng.uniform(1, 10, 40).round(1)
score = (45 + 4.5 * hours + rng.normal(0, 6, 40)).clip(0, 100).round(0)
df = pd.DataFrame({"hours": hours, "score": score})
print("Setup complete · ការរៀបចំបានបញ្ចប់. df has", len(df), "students.")
```

## Chapter 3 · Calculus for Optimisation
### ជំពូក ៣ · កាល់គុលសម្រាប់ការធ្វើឱ្យប្រសើរបំផុត

**Recap · រំលឹក**
- Derivative = slope: $f'(x) \approx \dfrac{f(x+h) - f(x-h)}{2h}$.
- Gradient $\nabla f$ collects partial derivatives; it points uphill.
- Gradient descent: $\mathbf{w} \leftarrow \mathbf{w} - \eta\,\nabla f(\mathbf{w})$.

ដេរីវេជាមេគុណប្រាប់ទិស។ ក្រាដ្យង់ចង្អុលទៅទិសកើនឡើង។ ការចុះតាមក្រាដ្យង់ដើរចុះក្រោមម្តងមួយជំហាន។

**Example · ឧទាហរណ៍**: run it and read the output. ដំណើរការ ហើយអានលទ្ធផល។

```python
def f(x):
    return x ** 2

def derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

print("f'(3) ≈", round(derivative(f, 3), 4))
```

## Chapter 4 · Probability
### ជំពូក ៤ · ប្រូបាប

**Recap · រំលឹក**
- $P(A \cup B) = P(A) + P(B) - P(A \cap B)$.
- Bayes: $P(A\mid B) = \dfrac{P(B\mid A)P(A)}{P(B)}$.
- `stats.binom(n, p)` and `stats.norm(mu, sigma)` give `.pmf`, `.cdf`, `.sf`.

ច្បាប់ប្រូបាប ទ្រឹស្តីបទបាយេស និងច្បាប់ទ្វេធា និងនរម៉ាល់។

```python
p_d, sens, fpr = 0.01, 0.95, 0.05
p_pos = sens * p_d + fpr * (1 - p_d)
print("P(disease | positive) =", round(sens * p_d / p_pos, 4))
print("P(Z > 1.96) =", round(stats.norm().sf(1.96), 4))
```

**Recap · រំលឹក**
- Standard error $SE = \sigma/\sqrt{n}$; the Central Limit Theorem makes sample means roughly normal.
- 95% confidence interval: $\bar{x} \pm t^{*}\, s/\sqrt{n}$ with `t_star = stats.t.ppf(0.975, n - 1)`.
- Two-sample test: `stats.ttest_ind(b, a, equal_var=False)`; reject $H_0$ if $p < 0.05$.
